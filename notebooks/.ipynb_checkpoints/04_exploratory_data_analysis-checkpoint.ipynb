## Project Objective

In this notebook, we will perform Exploratory Data Analysis (EDA) on the cleaned dataset.

The primary objective of this notebook is to:

- Understand customer interaction behavior
- Analyze user engagement patterns
- Explore item and category distributions
- Identify purchasing trends
- Detect behavioral insights useful for recommendation systems
- Generate business and machine learning insights from customer activity data

This notebook focuses strictly on exploratory analysis responsibilities following modular machine learning project architecture.

```python
# Data manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load cleaned datasets

events_df = pd.read_csv('../data/processed/cleaned_events.csv')

item_properties_df = pd.read_csv(
    '../data/processed/cleaned_item_properties.csv'
)

category_tree_df = pd.read_csv(
    '../data/processed/cleaned_category_tree.csv'
)
```

```python
print('Events Dataset Shape:', events_df.shape)
print('\n')

events_df.head()
```

```python
events_df.info()
```

```python
events_df.describe(include='all')
```

```python
# Event distribution
events_df['event'].value_counts()
```

```python
plt.figure(figsize=(8, 5))

events_df['event'].value_counts().plot(kind='bar')

plt.title('Customer Event Type Distribution')
plt.xlabel('Event Type')
plt.ylabel('Frequency')
plt.xticks(rotation=0)

plt.show()
```

### Observation-:
- View events dominate the dataset.
- Purchase events are comparatively lower.
- This indicates that user browsing behavior is significantly higher than completed transactions.

```python
# Unique users
unique_users = events_df['visitorid'].nunique()

print(f'Total Unique Users: {unique_users}')
```

```python
# Unique items
unique_items = events_df['itemid'].nunique()

print(f'Total Unique Items: {unique_items}')
```

```python
# Top active users
top_users = events_df['visitorid'].value_counts().head(10)

print(top_users)
```

```python
plt.figure(figsize=(10, 5))

top_users.plot(kind='bar')

plt.title('Top 10 Most Active Users')
plt.xlabel('Visitor ID')
plt.ylabel('Number of Interactions')

plt.show()
```
