```python
import pandas as pd

from sqlalchemy import text
from src.database import get_engine
```

```python
engine = get_engine()

print("Database connection established successfully.")
```

```python
events_df = pd.read_csv("../data/raw/events.csv")

item_prop_1_df = pd.read_csv("../data/raw/item_properties_part1.csv")

item_prop_2_df = pd.read_csv("../data/raw/item_properties_part2.csv")

category_df = pd.read_csv("../data/raw/category_tree.csv")
```

```python
print("Events Shape:", events_df.shape)

print("Item Properties Part 1 Shape:",
      item_prop_1_df.shape)

print("Item Properties Part 2 Shape:",
      item_prop_2_df.shape)

print("Category Tree Shape:",
      category_df.shape)
```

```python
# --------------------------------------------------
# Load Events Dataset
# --------------------------------------------------

print("Loading events data into MySQL...")

events_df.to_sql(
    name="events_raw",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=50000,
    method="multi"
)

print(f"Events Loaded: {len(events_df):,} rows")
```

```python
# --------------------------------------------------
# Load Item Properties Part 1
# --------------------------------------------------

print("Loading item_properties_part1 into MySQL...")

item_prop_1_df.to_sql(
    name="item_properties_raw",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=50000,
    method="multi"
)

print(f"Part 1 Loaded: {len(item_prop_1_df):,} rows")
```

```python
# --------------------------------------------------
# Load Item Properties Part 2
# --------------------------------------------------

print("Loading item_properties_part2 into MySQL...")

item_prop_2_df.to_sql(
    name="item_properties_raw",
    con=engine,
    if_exists="append",
    index=False,
    chunksize=50000,
    method="multi"
)

print(f"Part 2 Loaded: {len(item_prop_2_df):,} rows")
```

```python
# --------------------------------------------------
# Load Category Tree Dataset
# --------------------------------------------------

print("Loading category tree into MySQL...")

category_df.to_sql(
    name="category_tree_raw",
    con=engine,
    if_exists="replace",
    index=False
)

print(f"Category Tree Loaded: {len(category_df):,} rows")
```

```python
# --------------------------------------------------
# Validate Loaded Data
# --------------------------------------------------

from sqlalchemy import text

with engine.connect() as connection:

    events_count = connection.execute(
        text("SELECT COUNT(*) FROM events_raw")
    ).scalar()

    item_count = connection.execute(
        text("SELECT COUNT(*) FROM item_properties_raw")
    ).scalar()

    category_count = connection.execute(
        text("SELECT COUNT(*) FROM category_tree_raw")
    ).scalar()

print("=" * 50)
print("DATA LOAD COMPLETED")
print("=" * 50)

print(f"events_raw          : {events_count:,}")
print(f"item_properties_raw : {item_count:,}")
print(f"category_tree_raw   : {category_count:,}")
```
