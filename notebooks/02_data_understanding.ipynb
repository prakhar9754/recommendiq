```python
# Data manipulation libraries

import pandas as pd
import numpy as np

# Display settings

pd.set_option('display.max_columns', None)

print("Libraries imported successfully.")
```

```python
# Load datasets

events_df = pd.read_csv("../data/raw/events.csv")

item_prop_1_df = pd.read_csv("../data/raw/item_properties_part1.csv")

item_prop_2_df = pd.read_csv("../data/raw/item_properties_part2.csv")

category_tree_df = pd.read_csv("../data/raw/category_tree.csv")

print("Datasets loaded successfully.")
```

```python
# Check dataset shapes

print("Events Dataset Shape:", events_df.shape)

print("Item Properties Part 1 Shape:", item_prop_1_df.shape)

print("Item Properties Part 2 Shape:", item_prop_2_df.shape)

print("Category Tree Shape:", category_tree_df.shape)
```

```python
# Merge item properties datasets
item_properties_df = pd.concat(
    [item_prop_1_df, item_prop_2_df],
    axis=0,
    ignore_index=True
)

print('Combined Item Properties Shape:', item_properties_df.shape)
```

```python
item_properties_df.head()
```

```python
item_properties_df.info()
```

```python
item_properties_df.info()
```

```python
# Convert timestamp
item_properties_df['datetime'] = pd.to_datetime(
    item_properties_df['timestamp'],
    unit='ms'
)
```

```python
# Check duplicates
item_duplicates = item_properties_df.duplicated().sum()

print(f'Duplicate Rows: {item_duplicates}')
```

```python
# Remove duplicates
item_properties_df.drop_duplicates(inplace=True)
```

```python
category_tree_df.head()
```

```python
category_tree_df.info()
```

```python
# Missing parent categories
category_tree_df.isnull().sum()
```

```python
# Fill missing parentid(root id) with -1
category_tree_df['parentid'] = category_tree_df['parentid'].fillna(-1)
```

```python
print('Final Events Dataset Shape:', events_df.shape)
print('Null values in events_df:')
print(events_df.isnull().sum())
```

```python
print('Final Item Properties Shape:', item_properties_df.shape)
print('Null values in Item Properties:')
print(item_properties_df.isnull().sum())
```

```python
print('Final Category Tree Shape:', category_tree_df.shape)
print('Null values in category tree:')
print(category_tree_df.isnull().sum())
```

### Processed datasets are saved for feature engineering and model development

```python
# Save cleaned datasets

events_df.to_csv('../data/processed/cleaned_events.csv', index=False)

item_properties_df.to_csv(
    '../data/processed/cleaned_item_properties.csv',
    index=False
)

category_tree_df.to_csv(
    '../data/processed/cleaned_category_tree.csv',
    index=False
)
```
