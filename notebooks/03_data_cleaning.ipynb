## Duplicate Removal

Remove exact duplicate records where applicable.

```python
events_df.drop_duplicates(inplace=True)

item_properties_df.drop_duplicates(inplace=True)

category_tree_df.drop_duplicates(inplace=True)
```

## Timestamp Conversion

Convert Unix timestamps into datetime format for easier analysis.

```python
events_df["timestamp"] = pd.to_datetime(
    events_df["timestamp"],
    unit="ms"
)
```

```python
item_properties_df["timestamp"] = pd.to_datetime(
    item_properties_df["timestamp"],
    unit="ms"
)
```

## Column Standardization

Convert column names to lowercase for consistency.

```python
events_df.columns = (
    events_df.columns
    .str.strip()
    .str.lower()
)
```

```python
item_properties_df.columns = (
    item_properties_df.columns
    .str.strip()
    .str.lower()
)
```

```python
category_tree_df.columns = (
    category_tree_df.columns
    .str.strip()
    .str.lower()
)
```

## Save Cleaned Datasets

Store cleaned datasets in MySQL for downstream analysis.

```python
events_df.to_sql(
    "events_cleaned",
    engine,
    if_exists="replace",
    index=False
)
```

```python
item_properties_df.to_sql(
    "item_properties_cleaned",
    engine,
    if_exists="replace",
    index=False
)
```

```python
category_tree_df.to_sql(
    "category_tree_cleaned",
    engine,
    if_exists="replace",
    index=False
)
```

```python
from sqlalchemy import text

with engine.connect() as connection:

    events_count = connection.execute(
        text("SELECT COUNT(*) FROM events_cleaned")
    ).scalar()

    item_count = connection.execute(
        text(
            "SELECT COUNT(*) FROM item_properties_cleaned"
        )
    ).scalar()

    category_count = connection.execute(
        text(
            "SELECT COUNT(*) FROM category_tree_cleaned"
        )
    ).scalar()

print("events_cleaned:", events_count)

print("item_properties_cleaned:", item_count)

print("category_tree_cleaned:", category_count)
```
