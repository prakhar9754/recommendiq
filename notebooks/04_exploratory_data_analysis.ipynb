```python
# Data manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
from src.database import get_engine
engine = get_engine()
```

```python
events_df = pd.read_sql("SELECT * FROM events_cleaned LIMIT 500000",engine
)

item_properties_df = pd.read_sql(
    "SELECT * FROM item_properties_cleaned LIMIT 1000000",
    engine
)

category_tree_df = pd.read_sql(
    "SELECT * FROM category_tree_cleaned",
    engine
)
```

```python
print('Events Dataset Shape:', events_df.shape)
print('\n')

events_df.head()
```

```python
events_df.info()
```

```python
events_df.describe(include='all')
```

```python
# Event distribution
events_df['event'].value_counts()
```

```python
plt.figure(figsize=(8, 5))

events_df['event'].value_counts().plot(kind='bar')

plt.title('Customer Event Type Distribution')
plt.xlabel('Event Type')
plt.ylabel('Frequency')
plt.xticks(rotation=0)

plt.show()
```

### Observation-:
- View events dominate the dataset.
- Purchase events are comparatively lower.
- This indicates that user browsing behavior is significantly higher than completed transactions.

```python
# Unique users
unique_users = events_df['visitorid'].nunique()

print(f'Total Unique Users: {unique_users}')
```

```python
# Unique items
unique_items = events_df['itemid'].nunique()

print(f'Total Unique Items: {unique_items}')
```

```python
# Top active users
top_users = events_df['visitorid'].value_counts().head(10)

print(top_users)
```

```python
plt.figure(figsize=(10, 5))

top_users.plot(kind='bar')

plt.title('Top 10 Most Active Users')
plt.xlabel('Visitor ID')
plt.ylabel('Number of Interactions')

plt.show()
```

```python
top_items = events_df['itemid'].value_counts().head(10)
print('Most interacted items:')
print(top_items)
```

```python
plt.figure(figsize=(10, 5))

top_items.plot(kind='bar')

plt.title('Top 10 Most Interacted Items')
plt.xlabel('Item ID')
plt.ylabel('Interaction Count')

plt.show()
```

```python
# Convert datetime column
events_df['datetime'] = pd.to_datetime(events_df['timestamp'])
```

```python
events_df['month']  = events_df['datetime'].dt.month
```

```python
events_df['datetime'].min(), events_df['datetime'].max()
```

```python
monthly_interactions = events_df['month'].value_counts().sort_index()
print(monthly_interactions)
```

```python
plt.figure(figsize=(10, 5))

monthly_interactions.plot(kind='line', marker='o')

plt.title('Monthly Customer Interactions')
plt.xlabel('Month')
plt.ylabel('Interaction Count')

plt.show()
```

```python
events_df['hour'] = events_df['datetime'].dt.hour
```

```python
hourly_activity = events_df['hour'].value_counts().sort_index()
print(hourly_activity)
```

```python
plt.figure(figsize=(12, 5))

hourly_activity.plot(kind='line', marker='o')

plt.title('Hourly Customer Activity')
plt.xlabel('Hour of Day')
plt.ylabel('Interaction Count')

plt.show()
```

### Key Business Insights

- Customer browsing activity is significantly higher than purchase activity.
- A small group of users contributes to a large number of interactions.
- Certain products receive consistently high engagement.
- User activity varies across different hours and months.
- Interaction imbalance indicates the importance of implicit feedback in recommendation systems.



### Machine Learning Insights

- User-item interaction patterns can be used for collaborative filtering.
- Time-based interaction trends can improve personalization.
- Highly interacted items can contribute to popularity-based recommendations.
- Sparse purchase behavior suggests the need for behavioral recommendation strategies.
